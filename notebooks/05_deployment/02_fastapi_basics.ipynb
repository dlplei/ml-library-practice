## 1. 基本路由

```python
"""
basic_api.py - FastAPI 基础示例
"""

code = '''
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(
    title="ML API Demo",
    description="机器学习模型 API 演示",
    version="1.0.0"
)

# 基本路由
@app.get("/")
async def root():
    return {"message": "Hello World"}

@app.get("/health")
async def health_check():
    return {"status": "healthy"}

# 路径参数
@app.get("/items/{item_id}")
async def read_item(item_id: int):
    return {"item_id": item_id}

# 查询参数
@app.get("/users/")
async def read_users(skip: int = 0, limit: int = 10):
    return {"skip": skip, "limit": limit}
'''

print(code)
```

## 2. 请求体 (Pydantic)

```python
code = '''
from pydantic import BaseModel, Field

# 定义请求模型
class HouseFeatures(BaseModel):
    MedInc: float = Field(..., description="收入中位数", ge=0)
    HouseAge: float = Field(..., description="房屋年龄", ge=0)
    AveRooms: float = Field(..., description="平均房间数", ge=0)

class PredictionResponse(BaseModel):
    prediction: float
    features: dict

@app.post("/predict", response_model=PredictionResponse)
async def predict(house: HouseFeatures):
    # 模拟预测
    prediction = house.MedInc * 0.5 + house.HouseAge * 0.1
    
    return PredictionResponse(
        prediction=prediction,
        features=house.dict()
    )
'''

print(code)
```

## 3. 启动服务

```python
# 启动命令
print("""
# 开发模式（自动重载）
uvicorn main:app --reload --port 8000

# 生产模式
uvicorn main:app --host 0.0.0.0 --port 8000 --workers 4

# 访问 API 文档
# Swagger UI: http://localhost:8000/docs
# ReDoc: http://localhost:8000/redoc
""")
```

## 4. 测试 API

```python
# 使用 requests 测试
test_code = '''
import requests

# GET 请求
response = requests.get("http://localhost:8000/")
print(response.json())

# POST 请求
data = {
    "MedInc": 8.0,
    "HouseAge": 30.0,
    "AveRooms": 6.0
}
response = requests.post("http://localhost:8000/predict", json=data)
print(response.json())
'''

print(test_code)
```
