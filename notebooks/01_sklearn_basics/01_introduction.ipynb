## 1. 什么是机器学习？

机器学习是让计算机从数据中**自动学习规律**，而不需要显式编程的技术。

```
传统编程: 规则 + 数据 → 答案
机器学习: 答案 + 数据 → 规则 (模型)
```

## 2. 机器学习的三大范式

| 范式 | 描述 | 典型任务 | 示例算法 |
|------|------|----------|----------|
| **监督学习** | 有标签数据 | 分类、回归 | 线性回归、SVM、随机森林 |
| **无监督学习** | 无标签数据 | 聚类、降维 | KMeans、PCA、DBSCAN |
| **强化学习** | 奖励信号 | 决策、控制 | Q-Learning、PPO |

本项目主要关注**监督学习**。

## 3. scikit-learn 简介

scikit-learn 是 Python 最流行的传统机器学习库，特点：
- 统一的 API 设计
- 丰富的算法实现
- 完善的文档
- 与 NumPy/Pandas 无缝集成

````python
# 导入必要的库
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import sklearn
print(f'scikit-learn 版本: {sklearn.__version__}')

# 设置中文显示
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
````

## 4. scikit-learn 统一 API 设计

所有估计器 (Estimator) 都遵循相同的接口：

```python
# 1. 创建模型
model = Estimator(**params)

# 2. 训练 (fit)
model.fit(X_train, y_train)

# 3. 预测 (predict)
predictions = model.predict(X_test)

# 4. 评估
score = model.score(X_test, y_test)
```

````python
# 演示统一 API
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 1. 加载数据
iris = load_iris()
X, y = iris.data, iris.target

# 2. 划分数据集
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
print(f'训练集: {X_train.shape}, 测试集: {X_test.shape}')

# 3. 创建模型
model = RandomForestClassifier(n_estimators=100, random_state=42)

# 4. 训练
model.fit(X_train, y_train)

# 5. 预测
y_pred = model.predict(X_test)

# 6. 评估
accuracy = accuracy_score(y_test, y_pred)
print(f'\n准确率: {accuracy:.4f}')
````

## 5. 第一个实验：数据探索

让我们用鸢尾花数据集做一个简单的数据探索。

````python
# 数据探索
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
df['species'] = df['target'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

print('数据集概览:')
print(df.head())
print(f'\n数据集形状: {df.shape}')
print(f'\n各类别样本数:')
print(df['species'].value_counts())
````

````python
# 可视化 - 散点矩阵图
sns.pairplot(df, hue='species', palette='husl', diag_kind='kde')
plt.suptitle('鸢尾花数据集 - 散点矩阵图', y=1.02)
plt.show()
````

````python
# 相关性热力图
plt.figure(figsize=(8, 6))
corr = df[iris.feature_names].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('特征相关性矩阵')
plt.tight_layout()
plt.show()
````
