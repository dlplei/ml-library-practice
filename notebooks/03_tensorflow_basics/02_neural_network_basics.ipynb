```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

## 1. 激活函数

```python
x = np.linspace(-5, 5, 100)

activations = {
    'ReLU': tf.nn.relu(x).numpy(),
    'Sigmoid': tf.nn.sigmoid(x).numpy(),
    'Tanh': tf.nn.tanh(x).numpy(),
    'Leaky ReLU': tf.nn.leaky_relu(x, alpha=0.1).numpy(),
}

plt.figure(figsize=(12, 4))
for i, (name, y) in enumerate(activations.items(), 1):
    plt.subplot(1, 4, i)
    plt.plot(x, y, 'b-', linewidth=2)
    plt.title(name)
    plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. 简单神经网络

```python
# 手动实现简单神经网络
class SimpleNN:
    def __init__(self, input_size, hidden_size, output_size):
        self.W1 = tf.Variable(tf.random.normal([input_size, hidden_size]) * 0.01)
        self.b1 = tf.Variable(tf.zeros([hidden_size]))
        self.W2 = tf.Variable(tf.random.normal([hidden_size, output_size]) * 0.01)
        self.b2 = tf.Variable(tf.zeros([output_size]))
    
    def __call__(self, x):
        self.z1 = tf.matmul(x, self.W1) + self.b1
        self.a1 = tf.nn.relu(self.z1)
        self.z2 = tf.matmul(self.a1, self.W2) + self.b2
        return tf.nn.softmax(self.z2)

# 测试
nn = SimpleNN(4, 10, 3)
x_test = tf.random.normal([5, 4])
output = nn(x_test)
print(f'输入形状: {x_test.shape}')
print(f'输出形状: {output.shape}')
print(f'输出:\n{output.numpy()}')
```

## 3. XOR 问题

```python
# XOR 数据
X = tf.constant([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=tf.float32)
y = tf.constant([[0], [1], [1], [0]], dtype=tf.float32)

# 训练
model = SimpleNN(2, 4, 1)
optimizer = tf.optimizers.Adam(learning_rate=0.1)

for epoch in range(1000):
    with tf.GradientTape() as tape:
        pred = model(X)
        loss = tf.reduce_mean(tf.square(pred - y))
    
    grads = tape.gradient(loss, [model.W1, model.b1, model.W2, model.b2])
    optimizer.apply_gradients(zip(grads, [model.W1, model.b1, model.W2, model.b2]))
    
    if (epoch + 1) % 200 == 0:
        print(f'Epoch {epoch+1}, Loss: {loss.numpy():.4f}')

print(f'\n预测结果:')
for i in range(4):
    print(f'  {X[i].numpy()} -> {model(X[i:i+1]).numpy()[0][0]:.4f}')
```
