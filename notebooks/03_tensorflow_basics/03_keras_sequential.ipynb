# 03 - Keras Sequential API

## 学习目标
- 掌握 Keras Sequential API
- 构建第一个神经网络
- MNIST 手写数字识别

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
import numpy as np
```

## 1. 加载 MNIST 数据

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

print(f'训练集: {x_train.shape}')
print(f'测试集: {x_test.shape}')

# 归一化
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# 可视化
plt.figure(figsize=(10, 5))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title(f'Label: {y_train[i]}')
    plt.axis('off')
plt.tight_layout()
plt.show()
```

## 2. 构建模型

```python
model = keras.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

## 3. 训练模型

```python
history = model.fit(
    x_train, y_train,
    validation_split=0.1,
    epochs=15,
    batch_size=128,
    verbose=1
)
```

## 4. 可视化训练过程

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(history.history['accuracy'], label='训练准确率')
ax1.plot(history.history['val_accuracy'], label='验证准确率')
ax1.set_title('模型准确率')
ax1.set_xlabel('Epoch')
ax1.legend()

ax2.plot(history.history['loss'], label='训练损失')
ax2.plot(history.history['val_loss'], label='验证损失')
ax2.set_title('模型损失')
ax2.set_xlabel('Epoch')
ax2.legend()

plt.tight_layout()
plt.show()
```

## 5. 评估与预测

```python
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f'\n测试集准确率: {test_acc:.4f}')

# 预测
predictions = model.predict(x_test[:10])

plt.figure(figsize=(15, 5))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(x_test[i], cmap='gray')
    pred_label = np.argmax(predictions[i])
    color = 'green' if pred_label == y_test[i] else 'red'
    plt.title(f'Pred: {pred_label}, True: {y_test[i]}', color=color)
    plt.axis('off')
plt.tight_layout()
plt.show()
```

## 小结
- Sequential API 简单直观
- Dropout 防止过拟合
- 训练曲线帮助诊断问题
