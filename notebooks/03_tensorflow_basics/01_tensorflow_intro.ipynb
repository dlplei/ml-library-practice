# 01 - TensorFlow 基础与张量操作

## 学习目标
- 理解 TensorFlow 的核心概念
- 掌握张量操作
- 了解自动微分

```python
import tensorflow as tf
import numpy as np

print(f'TensorFlow 版本: {tf.__version__}')
print(f'GPU 可用: {tf.config.list_physical_devices("GPU")}')
```

## 1. 张量创建

```python
# 常量张量
a = tf.constant([1, 2, 3, 4, 5])
print(f'常量: {a}')

# 从 NumPy 创建
b = tf.constant(np.array([[1, 2], [3, 4]]))
print(f'\n2D 张量:\n{b}')

# 特殊张量
zeros = tf.zeros([3, 3])
ones = tf.ones([2, 4])
random = tf.random.normal([3, 3])

print(f'\nZeros:\n{zeros}')
print(f'\nRandom:\n{random}')
```

## 2. 张量运算

```python
x = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)
y = tf.constant([[5, 6], [7, 8]], dtype=tf.float32)

# 基本运算
print(f'加法:\n{x + y}')
print(f'\n乘法 (元素级):\n{x * y}')
print(f'\n矩阵乘法:\n{tf.matmul(x, y)}')

# 归约操作
print(f'\n求和: {tf.reduce_sum(x)}')
print(f'均值: {tf.reduce_mean(x)}')
print(f'最大值: {tf.reduce_max(x)}')
```

## 3. 自动微分

```python
# 计算 f(x) = x^2 在 x=3 处的导数
x = tf.Variable(3.0)

with tf.GradientTape() as tape:
    y = x ** 2

dy_dx = tape.gradient(y, x)
print(f'f(x) = x² 在 x=3 处的导数: {dy_dx.numpy()}')  # 应该是 6

# 更复杂的例子
w = tf.Variable(2.0)
b = tf.Variable(1.0)

with tf.GradientTape() as tape:
    x_val = tf.constant(3.0)
    y_pred = w * x_val + b
    loss = (y_pred - 10.0) ** 2  # 目标值是 10

grads = tape.gradient(loss, [w, b])
print(f'\n损失: {loss.numpy()}')
print(f'梯度 w: {grads[0].numpy()}')
print(f'梯度 b: {grads[1].numpy()}')
```

## 4. Variable

```python
# Variable 是可变的
v = tf.Variable([1.0, 2.0, 3.0])
print(f'初始值: {v.numpy()}')

v.assign([4.0, 5.0, 6.0])
print(f'assign 后: {v.numpy()}')

v.assign_add([1.0, 1.0, 1.0])
print(f'assign_add 后: {v.numpy()}')
```

## 小结
- Tensor 是不可变的多维数组
- Variable 是可变的，用于模型参数
- GradientTape 实现自动微分
