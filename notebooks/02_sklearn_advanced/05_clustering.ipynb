# 05 - 聚类算法

## 学习目标
- 掌握 KMeans、DBSCAN、层次聚类
- 学会评估聚类效果

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_moons
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.metrics import silhouette_score, silhouette_samples
from sklearn.preprocessing import StandardScaler
```

## 1. KMeans

```python
# 生成数据
X, y_true = make_blobs(n_samples=300, centers=4, random_state=42, cluster_std=0.8)

# KMeans
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
y_pred = kmeans.fit_predict(X)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(X[:, 0], X[:, 1], c=y_true, cmap='viridis', s=30)
axes[0].set_title('真实标签')
axes[1].scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis', s=30)
axes[1].scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
               c='red', marker='x', s=200, linewidths=3)
axes[1].set_title('KMeans 聚类')
plt.tight_layout()
plt.show()
```

## 2. 肘部法则

```python
inertias = []
silhouettes = []
K = range(2, 11)

for k in K:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X, km.labels_))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.plot(K, inertias, 'bo-')
ax1.set_xlabel('K')
ax1.set_ylabel('Inertia')
ax1.set_title('肘部法则')

ax2.plot(K, silhouettes, 'ro-')
ax2.set_xlabel('K')
ax2.set_ylabel('Silhouette Score')
ax2.set_title('轮廓系数')
plt.tight_layout()
plt.show()
```

## 3. DBSCAN

```python
# 月牙形数据
X_moons, y_moons = make_moons(n_samples=300, noise=0.05, random_state=42)

# DBSCAN
dbscan = DBSCAN(eps=0.2, min_samples=5)
y_dbscan = dbscan.fit_predict(X_moons)

plt.figure(figsize=(10, 5))
plt.scatter(X_moons[:, 0], X_moons[:, 1], c=y_dbscan, cmap='viridis', s=30)
plt.title(f'DBSCAN (eps=0.2, min_samples=5), 聚类数: {len(set(y_dbscan)) - (1 if -1 in y_dbscan else 0)}')
plt.show()
```

## 小结
- KMeans 适合球形簇
- DBSCAN 可以发现任意形状的簇
- 轮廓系数评估聚类质量
